```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
amplitude = 1  # Volts
frequency = 100  # Hz
sampling_rate = 1000  # Hz (number of samples per second)

duration = 0.02  # Seconds (e.g., 2 cycles at 100 Hz)
Fs = 100000

# Time vector
t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)

# Generate sine wave
sine_wave = amplitude * np.sin(2 * np.pi * frequency * t)

# Plot the sine wave
plt.figure(figsize=(10, 4))
plt.plot(t, sine_wave)
plt.title('Sine Wave (1V Amplitude, 100Hz Frequency)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude (V)')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

# Parameters
amplitude = 1  # Volts
frequency = 100  # Hz
sampling_rate = 10000  # Hz (number of samples per second)
duration = 0.02  # Seconds

# Time vector
t = np.linspace(0, duration, int(sampling_rate * duration), endpoint=False)

# Generate square wave
square_wave = amplitude * signal.square(2 * np.pi * frequency * t)

# Plot the square wave
plt.figure(figsize=(10, 4))
plt.plot(t, square_wave)
plt.title('Square Wave (1V Amplitude, 100Hz Frequency)')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude (V)')
plt.ylim(-1.2, 1.2) # Set y-axis limits for better visualization
plt.grid(True)
plt.show()
```
